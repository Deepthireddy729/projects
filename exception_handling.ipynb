```python
name = input("What's your name? ")

file = open("names.txt", "w")
file.write(name)
file.close()

names = []

for _ in range(3):
    names.append(input("What's your name?" ))

for name in sorted(names):
    print(f"Hello, {name}")
```

```python
#READ FROM FILE:
#read:reads the entire content of file
#readline:reads one line from file at a time
#readlines:reads all lines into the list
file = open("names.txt", "r")
content=file.read()
print(content)
file.close()


file = open("names.txt", "r")
line=file.readline()
print(line)
file.close()
```

```python
#WRITE:

#write:Writes a string to a file.
#writeline:writes a list of strings to a file
file = open("names.txt", "w")
file.write("Harry\n")
file.write("Ron\n")
file.write("Hermione\n")
file.close()
```
