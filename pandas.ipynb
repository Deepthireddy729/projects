```python
import pandas as pd
#Pandas is built on top of NumPy and provides high-level data structures like Series and DataFrame for fast and efficient data manipulation.
```

```python
df = pd.read_csv('/content/sample_data/california_housing_test.csv')
print(df.head())

```

```python
# Check for missing values
print(df.isnull().sum())
print(df.columns)

df.dropna(inplace=True)
```

```python
# Check for duplicate rows
print(df.duplicated().sum())
```

```python
# Drop duplicate rows
df.drop_duplicates(inplace=True)
```

```python
#Checking Data Types: Use dtypes to see the data type of each column.
#Converting Data Types: Use astype() to convert data types.
print(df.dtypes)

# Convert a column to numeric
df['latitude'] = pd.to_numeric(df['latitude'], errors='coerce') # Placeholder - causes KeyError

# Convert a column to datetime
df['date_column'] = pd.to_datetime(df['date_column']) # Placeholder - causes KeyError
```

```python
 #Filter rows where 'age' is greater than 30
filtered_df = df[df['latitude'] < 30]

# Sort the DataFrame by 'age' in descending order
sorted_df = df.sort_values('housing_median_age', ascending=False)
print(filtered_df)
print(sorted_df)
```

```python
#Pandas makes it easy to identify and remove outliers based on statistical methods like the interquartile range (IQR).
Q1 = df['median_house_value'].quantile(0.25)
Q3 = df['median_house_value'].quantile(0.75)
IQR = Q3 - Q1

df = df[~(
    (df['median_house_value'] < (Q1 - 1.5 * IQR)) |
    (df['median_house_value'] > (Q3 + 1.5 * IQR))
)]

print(df)
```

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'City': ['New York', 'Chicago', 'New York', 'San Francisco']}

df = pd.DataFrame(data)

# Filtering rows where Age is greater than 30
filtered_df = df[df['Age'] > 30]
print(filtered_df)
```

```python
# Filtering with multiple conditions (AND condition)
filtered_df = df[(df['Age'] > 30) & (df['City'] == 'New York')]
print(filtered_df)
```

```python
# Filtering rows where Age > 30 OR City is New York
filtered_df = df[(df['Age'] > 30) | (df['City'] == 'New York')]
print(filtered_df)
```

```python
# Exclude rows where City is 'Chicago'
filtered_df = df[~(df['City'] == 'Chicago')]
print(filtered_df)
```

```python
# Using .query() method
filtered_df = df.query("Age > 30 and City == 'New York'")
print(filtered_df)
```

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, 35, 40, 29],
        'City': ['New York', 'Chicago', 'New York', 'San Francisco', 'Los Angeles']}

df = pd.DataFrame(data)

# Filtering rows where City is either 'New York' or 'San Francisco'
filtered_df = df[df['City'].isin(['New York', 'San Francisco'])]
print(filtered_df)
```

```python
# Excluding rows where City is 'New York' or 'San Francisco'
filtered_df = df[~df['City'].isin(['New York', 'San Francisco'])]
print(filtered_df)
```

```python
# Filtering rows where Name starts with 'A'
filtered_df = df[df['Name'].str.contains('A', case=False)]  # case=False makes it case-insensitive
print(filtered_df)
```

```python
# Filtering rows where City contains 'York'
filtered_df = df[df['City'].str.contains('York', case=False)]
print(filtered_df)
```

```python
# Using .loc for filtering and selecting specific columns
#More readable than df[...] for selecting columns.
#Faster execution when working with large datasets.
#Great for clean, structured data extraction.
filtered_df = df.loc[df['Age'] > 30, ['Name', 'City']]
print(filtered_df)
```

```python
# Selecting only columns with 'Name' or 'Age'
filtered_df = df.filter(items=['Name', 'Age'])
print(filtered_df)
```
