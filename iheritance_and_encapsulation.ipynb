```python
#SINGLE INHERITANCE:
class Animal:#single inheritance which inherits the properties of parent class to child class where only one parent and child exists
    def eat(self):#eat is a method
        print("This animal can eat.")

class Dog(Animal):#it is using animal class properties
    def bark(self):#child class added another method
        print("The dog barks: Woof woof!")

dog = Dog()#object created for dog class
dog.eat()  #where it can get properties of animal class
dog.bark()
```

```python
#MULTIPLE INHERITANCE:where a class inherits properties and methods from more than one parent class
class A:
    def feature1(self):#feature 1 is method from class A
        print("Feature 1 from class A")

class B:
    def feature2(self):#feature 2 is method from class B
        print("Feature 2 from class B")

class C(A, B):#class c is child class which inherits properties from both class A and class B
    pass#it does nothing

obj = C()#object has been created for class C
obj.feature1()  #properties of class A
obj.feature2()  #properties of class B
#we can even create an additional method in child class if required
```

```python
#MULTILEVEL INHERITANCE:A class inherits from another class, which itself inherits from another class. It creates an inheritance chain.
class Grandparent:
    def feature1(self):
        print("Feature 1 from Grandparent")

class Parent(Grandparent):#it inherits properties of grandparent class
    def feature2(self):
        print("Feature 2 from Parent")

class Child(Parent):#it can inherit properties from both parent and grandparent class
    def feature3(self):
        print("Feature 3 from Child")

obj = Child()
obj.feature1()
obj.feature2()
obj.feature3()
```

```python
#heirarchial inheritance:one parent class with multiple child classes.
class Student:#Student is parent rest all are child classes which inherits from Student
    def study(self):
        print("This student studies.")

class MathStudent(Student):
    def solve_equations(self):
        print("Solving math equations.")

class ArtStudent(Student):
    def draw(self):
        print("Drawing a masterpiece.")

math_student = MathStudent()
math_student.study()
math_student.solve_equations()

art_student = ArtStudent()
art_student.study()
art_student.draw()
```

```python
#Inheritance allows one class (child) to reuse and extend the properties and methods of another class (parent).
#single inheritance which inherits the properties of parent class to child class where only one parent and child exists
#MULTIPLE INHERITANCE:where a class inherits properties and methods from more than one parent class
#MULTILEVEL INHERITANCE:A class inherits from another class, which itself inherits from another class. It creates an inheritance chain.
#heirarchial inheritance:one parent class with multiple child classes.
```

```python
#data (attributes) and the methods that manipulate it into a single unit — a class — while restricting direct access to the data.
#Public: No prefix, accessible everywhere (e.g., data).
#Protected: Single underscore (_data),a hint for internal or subclass use, though not enforced.
#Private: Double underscore (__data), triggering name mangling to discourage direct access.
```

```python
class WeatherStation:
    def __init__(self, temperature):
        self.temp = temperature  # Public attribute

    def report(self):
        print(f"Current temperature: {self.temp}°C")

# Using the class
station = WeatherStation(25)
print(station.temp)
station.temp = -500
print(station.temp)
station.report()
```

```python
class Inventory:
    def __init__(self, stock):
        self._stock = stock  # Protected attribute

    def check_stock(self):
        print(f"Items in stock: {self._stock}")

    def restock(self, amount):
        self._stock += amount
        print(f"Restocked {amount}. New stock: {self._stock}")

# Using the class
shop = Inventory(100)
print(shop._stock)  # Accessible, but discouraged
shop._stock = -10  # Can still modify, though it’s "protected"
print(shop._stock)
shop.check_stock()
shop.restock(50)
```
