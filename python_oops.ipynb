```python
class Number:
    def __init__(self, n):
        self.n = n

    def check(self):
        if self.n % 2 == 0:
            print("Even")
        else:
            print("Odd")

n = Number(10)
n.check()
```

```python
class Math:
    def __init__(self, n):
        self.n = n

    def square(self):
        print(self.n * self.n)

m = Math(5)
m.square()
```

```python
class Rectangle:
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        print(self.l * self.b)

r = Rectangle(4, 5)
r.area()
```
